```python
def cscan(requests, head, disk_size):
    requests = sorted(requests)
    total_movement = 0
    current_head = head
    sequence = []

    # Split requests into two parts
    right = [r for r in requests if r >= head]
    left = [r for r in requests if r < head]

    # Process in one direction
    for request in right:
        total_movement += abs(request - current_head)
        sequence.append(request)
        current_head = request

    # Jump to the beginning of the disk
    total_movement += abs(disk_size - 1 - current_head) + (disk_size - 1)
    current_head = 0

    # Process the other direction
    for request in left:
        total_movement += abs(request - current_head)
        sequence.append(request)
        current_head = request

    print("C-SCAN Sequence:", sequence)
    print("Total Head Movement:", total_movement)

# Example Usage
cscan([98, 183, 37, 122, 14, 124, 65, 67], head=53, disk_size=200)
```
