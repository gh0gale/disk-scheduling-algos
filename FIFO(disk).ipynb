```python
def fifo(requests, head):
    total_movement = 0
    current_head = head
    sequence = []

    for request in requests:
        total_movement += abs(request - current_head)
        sequence.append(request)
        current_head = request

    print("FIFO Sequence:", sequence)
    print("Total Head Movement:", total_movement)

# Example Usage
fifo([98, 183, 37, 122, 14, 124, 65, 67], head=53)
```
