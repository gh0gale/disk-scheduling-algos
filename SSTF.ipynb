```python
def sstf(requests, head):
    total_movement = 0
    current_head = head
    sequence = []
    requests = requests[:]

    while requests:
        closest_request = min(requests, key=lambda r: abs(r - current_head))
        total_movement += abs(closest_request - current_head)
        sequence.append(closest_request)
        current_head = closest_request
        requests.remove(closest_request)

    print("SSTF Sequence:", sequence)
    print("Total Head Movement:", total_movement)

# Example Usage
sstf([98, 183, 37, 122, 14, 124, 65, 67], head=53)
```
